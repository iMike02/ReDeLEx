```python
# How to open and analyze the CSV file created by compute_graph_metrics.py
import pandas as pd
pd.set_option('display.max_columns', 100)
pd.set_option('display.max_rows', 100)

csv_path = '/home/gabrimi8/RDL/ReDeLEx/graph_metrics_test_hubs_bridges.csv'
df = pd.read_csv(csv_path)

print("\nAvailable columns:")
print(df.columns.tolist())
# df.head()
```

```python

# Find top 5 datasets for a specific metric
metric_column = 'delta_diameter'
top_5_highest = df.nlargest(5, metric_column)[['dataset', metric_column]]
print(f"\nTop 5 datasets with HIGHEST {metric_column}:")
print(top_5_highest.to_string(index=False))

top_5_lowest = df.nsmallest(5, metric_column)[['dataset', metric_column]]
print(f"\nTop 5 datasets with LOWEST {metric_column}:")
print(top_5_lowest.to_string(index=False))

```
